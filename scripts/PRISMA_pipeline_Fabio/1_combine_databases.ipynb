```python
import os
import glob
import pandas as pd
# Mount Google Drive
from google.colab import drive as gdrive
```

```python
# Define the folder containing the subfolders with CSV files
main_folder = '/content/drive/MyDrive/[ICON fellowship] social robot and autism/AI for behavioral analysis and autism/queries'
```

```python
# Define the target CSV file name
target_csv_name = 'autism_formatted_controlled.csv'

# Get a list of all subfolders
subfolders = [f.path for f in os.scandir(main_folder) if f.is_dir()]

# Initialize an empty DataFrame to store combined data
combined_data = pd.DataFrame()

# Loop through each subfolder
for subfolder in subfolders:
    # List all files in the subfolder
    files_in_subfolder = os.listdir(subfolder)

    # Filter out CSV files
    csv_files = [file for file in files_in_subfolder if file.endswith('.csv')]

    # Iterate through CSV files in the subfolder
    for csv_file in csv_files:
        if csv_file == target_csv_name:
          # Construct full path to the CSV file
          csv_path = os.path.join(subfolder, csv_file)
          print(csv_path)

          # Read CSV file into a DataFrame
          data = pd.read_csv(csv_path)
          print(data.columns)
          # Add an extra column "database" with the subfolder name
          data['Database'] = [os.path.basename(subfolder).replace("(new method)", "")] * len(data)

          # Append data to the combined DataFrame
          combined_data = combined_data.append(data, ignore_index=True)
```

```python
# Save the combined DataFrame to the target CSV file
combined_data.to_csv(f'{main_folder}/autism_formatted__controlled_combined.csv', index=False)
```
