```python
# Mount Google Drive
from google.colab import drive as gdrive
import os
import csv
import pandas as pd
import pandas as pd
from collections import OrderedDict
```

```python
workspace = "/content/drive/MyDrive/[ICON fellowship] social robot and autism/AI for behavioral analysis and autism/queries/PubMed"
```

```python
# Folder containing .bib file
csv_folder = f"{workspace}/csv/"

# Output CSV file
output_csv = f"{workspace}/autism.csv"

output_csv_formatted = f"{workspace}/autism_formatted.csv"
```

```python
# Get a list of all CSV files in the folder
csv_files = [file for file in os.listdir(csv_folder) if file.endswith('.csv')]

# Check if there are any CSV files in the folder
if len(csv_files) == 0:
    print("No CSV files found in the folder.")
else:
    # Initialize an empty DataFrame to store the merged data
    merged_data = pd.DataFrame()

    # Loop through the CSV files and merge them into the DataFrame
    for csv_file in csv_files:
        file_path = os.path.join(csv_folder, csv_file)
        print(file_path)
        df = pd.read_csv(file_path)
        merged_data = merged_data.append(df, ignore_index=True)

    # Save the merged data to the output CSV file
    merged_data.to_csv(output_csv, index=False)

    print(f"Merged data saved to {output_csv}")
```

```python
df = pd.read_csv(output_csv)
df
```

```python
df.columns
```

```python
df['Abstract'] = None
df['Keywords'] = None
df['URL'] = None
```

```python
# Define a dictionary to map the old column names to the new ones
column_mapping = OrderedDict([
    ('Title', 'Title'),
    ('Abstract', 'Abstract'),
    ('Keywords', 'Keywords'),
    ('DOI', 'DOI'),
    ('URL', 'URL'),
    ('Authors', 'Authors'),
    ('Journal/Book', 'Venue'),
    ('Publication Year', 'Year')
])

column_order = ['Title', 'Abstract', 'Keywords', 'DOI', 'URL', 'Authors', 'Venue', 'Year']

# Rename the columns based on the mapping
df.rename(columns=column_mapping, inplace=True)

# List of columns to drop
columns_to_drop = [col for col in df.columns if col not in column_mapping.values()]

# Drop the unwanted columns
df.drop(columns=columns_to_drop, inplace=True)

df = df[column_order]
```

```python
df.to_csv(output_csv_formatted, index=False)
```
