```python
# Define the target CSV file name
target_csv_name = 'autism_formatted.csv'

# Get a list of all subfolders
subfolders = [f.path for f in os.scandir(main_folder) if f.is_dir()]

# Loop through each subfolder
for subfolder in subfolders:
    # List all files in the subfolder
    files_in_subfolder = os.listdir(subfolder)

    # Filter out CSV files
    csv_files = [file for file in files_in_subfolder if file.endswith('.csv')]

    # Iterate through CSV files in the subfolder
    for csv_file in csv_files:
        if csv_file == target_csv_name:
          # Construct full path to the CSV file
          csv_path = os.path.join(subfolder, csv_file)
          print(csv_path)

          # Read CSV file into a DataFrame
          df = pd.read_csv(csv_path)

          df['Title'].fillna("", inplace=True)
          df['Abstract'].fillna("", inplace=True)
          df['Keywords'].fillna("", inplace=True)

          for filter_words in [autism_words]:
            print(len(df))
            df = df[
                df['Title'].str.contains('|'.join(fr'\b{word}\b' for word in filter_words), case=False) |
                df['Abstract'].str.contains('|'.join(fr'\b{word}\b' for word in filter_words), case=False) |
                df['Keywords'].str.contains('|'.join(fr'\b{word}\b' for word in filter_words), case=False)
            ]
            print(len(df))

          # Replace empty strings with None in the 'Title' column
          df.loc[df['Title'] == "", 'Title'] = None
          # Replace empty strings with None in the 'Abstract' column
          df.loc[df['Abstract'] == "", 'Abstract'] = None
          # Replace empty strings with None in the 'Keywords' column
          df.loc[df['Keywords'] == "", 'Keywords'] = None

          df = df.reset_index(drop=True)
          df.to_csv(csv_path.replace(".csv", "_controlled.csv"), index=False)
```
