```python
# Define a function to update word counts in the dictionary
def update_word_counts(text, word_count_dict):
    if text is not None:
        text = f"{text}".lower()
        for word in word_count_dict:
            # word_count_dict[word].append(text.count(word.lower()))
            if is_word_in_text(word, text):
                word_count_dict[word].append(count_word_in_text(word, text))
            else:
                word_count_dict[word].append(0)
    else:
        for word in word_count_dict:
            word_count_dict[word].append(0)

def count_word_in_text(word, text):
    # Create a regular expression pattern to match whole words
    pattern = r'\b{}\b'.format(re.escape(word))

    # Use re.findall() to find all occurrences of the word as a whole word
    matches = re.findall(pattern, text)

    # Return the count of matches
    return len(matches)

def is_word_in_text(word, text):
    # Create a regular expression pattern to match whole words
    pattern = r'\b{}\b'.format(re.escape(word))

    # Search for the word in the text using the pattern
    match = re.search(pattern, text)

    # If a match is found, return True; otherwise, return False
    return bool(match)

# Define a custom function to check if a word appears alone in the text
def is_alone_in_text(text, word, other_words):
    text = text.lower()
    word = word.lower()
    other_words = [w.lower() for w in other_words]

    # Check if the word is in the text
    if not is_word_in_text(word, text):
        return False

    # Check if none of the other words are in the text
    return all(not is_word_in_text(other_word, text) for other_word in other_words)

# Define a custom function to concatenate the columns
def concatenate_columns(row):
    return ' '.join(filter(None, [f"{row['Title']}", f"{row['Abstract']}", f"{row['Keywords']}"]))

def run_stats(df, ref_words):
  # Initialize a dictionary to store the word counts as lists
  word_counts = {word: [] for word in ref_words}

  # Apply the function to create the utility column
  df['Utility'] = df.apply(concatenate_columns, axis=1)
  df['Utility'].apply(update_word_counts, args=(word_counts,))

  # Initialize a dictionary to store the counts
  single_word_counts = {word: 0 for word in ref_words}

  # Count the times each word appears alone
  for word in ref_words:
      other_words = [w for w in ref_words if w != word]
      single_word_counts[word] = df['Utility'].apply(lambda text: is_alone_in_text(text, word, other_words)).sum()

  '''
  # Print the counts
  for word, count in single_word_counts.items():
      print(f'"{word}": {count}')
  '''

  # Print the word counts as lists
  for word, counts in word_counts.items():
      print(f'Summary statistics for {word}:')
      print(f'Sum: {sum(counts)}')
      print(f'Mean: {statistics.mean(counts)}')
      print(f'Standard Deviation: {statistics.stdev(counts)}')
      print(f'Minimum: {min(counts)}')
      print(f'Maximum: {max(counts)}')
      print(f'Single word: {single_word_counts[word]} times')
      print()
```
