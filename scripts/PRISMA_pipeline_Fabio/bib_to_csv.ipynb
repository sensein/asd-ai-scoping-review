```python
# Mount Google Drive
from google.colab import drive as gdrive
import os
import csv
from bibtexparser.bparser import BibTexParser
import pandas as pd
from collections import OrderedDict
```

```python
gdrive.mount('/content/drive')
```

```python
workspace = "/content/drive/MyDrive/[ICON fellowship] social robot and autism/AI for behavioral analysis and autism/queries/acm_digital_library (new method)"
```

```python
# Folder containing .bib file
bib_folder = f"{workspace}/bib/"

# Output CSV file
output_csv = f"{workspace}/autism.csv"

output_csv_formatted = f"{workspace}/autism_formatted.csv"
```

```python
# Set to store all unique fields encountered
all_fields = set()

# List to store all entries
all_entries = []

# Loop through each .bib file in the folder
for bib_file in os.listdir(bib_folder):
    if bib_file.endswith(".bib"):
        bib_path = os.path.join(bib_folder, bib_file)
        print(bib_path)

        # Parse the .bib file
        with open(bib_path, "r", encoding="utf-8") as file:
            parser = BibTexParser()
            parser.ignore_nonstandard_types = False
            bib_data = parser.parse_file(file)

        # Extract and store each entry
        for entry in bib_data.entries:
            entry_dict = {"Key": entry.get("ID")}
            entry_dict.update(entry)
            all_entries.append(entry_dict)
            all_fields.update(entry.keys())  # Update the set of all encountered fields

print(len(all_entries))

# Convert the set of all fields to a sorted list
fieldnames = ["Key"] + sorted(all_fields)

# Write entries to CSV
with open(output_csv, "w", newline="", encoding="utf-8") as csv_file:
    csv_writer = csv.DictWriter(csv_file, fieldnames=fieldnames)
    csv_writer.writeheader()
    csv_writer.writerows(all_entries)

print(f"CSV file '{output_csv}' created with {len(all_entries)} entries.")
```

```python
df = pd.read_csv(output_csv)
df
```

```python
df = df.drop_duplicates(subset=['ID'])
df
```

```python
# Define a dictionary to map the old column names to the new ones
column_mapping = OrderedDict([
    ('title', 'Title'),
    ('abstract', 'Abstract'),
    ('keywords', 'Keywords'),
    ('doi', 'DOI'),
    ('url', 'URL'),
    ('author', 'Authors'),
    ('booktitle', 'Venue'),
    ('year', 'Year')
])

column_order = ['Title', 'Abstract', 'Keywords', 'DOI', 'URL', 'Authors', 'Venue', 'Year']

# Rename the columns based on the mapping
df.rename(columns=column_mapping, inplace=True)

# List of columns to drop
columns_to_drop = [col for col in df.columns if col not in column_mapping.values()]

# Drop the unwanted columns
df.drop(columns=columns_to_drop, inplace=True)

df = df[column_order]
```
